# 実験結果の可視化

`outputs/` 以下に保存された実験結果を読み込み，図を `outputs/{実験名}/figures/` に保存する．
複数のseedがある条件は，seed間の平均と標準偏差を計算して描画する．

現在対応している実験:

- `ex001_vae_latent_diagnosis`: VAE潜在空間のサンプリング挙動の診断 (実験1〜3)
- `ex002_partial_prior_sampling`: 属性次元を保持し，他の次元を事前分布から取るサンプリングの検証 (実験A・B)
- `ex003_ghost_image_eval`: GHOSTで生成した画像による評価 (作業前の確認，本実験，貼り戻した画像での評価)

```python
import os
import sys

REPO_ROOT = os.getcwd()
sys.path.insert(0, os.path.join(REPO_ROOT, "programs", "ex001_vae_latent_diagnosis"))

from IPython.display import Image, display

import common
import plots
```

## 可視化する条件の指定

可視化したい実験・seed・部分集合を以下の変数で指定する．

```python
# 可視化する実験 ("ex1", "ex2", "ex3" から選択)
EXPERIMENTS = ["ex1", "ex2", "ex3"]

# 平均と標準偏差の計算に用いるseed
SEEDS = [0, 1, 2]

# サンプル数 K を表すハイパーパラメータ (outputs/{実験名}/{手法}/{ここ}/{seed}/)
HYPERPARAMETER = f"K{common.NUM_SAMPLES}"

# 実験2で集計する部分集合 ("all": 全画像, "train": VAEの学習用, "test": VAEの検証用)
EX2_SUBSETS = ["all", "train", "test"]

# 図をノートブック上にも表示するかどうか
SHOW_INLINE = True
```

```python
def show(paths):
    """
    保存した図のパスを出力し，必要に応じてノートブック上に表示する．

    引数:
        paths (list[str]): 図のパスのリスト．
    戻り値:
        なし．
    """
    for path in paths:
        print(os.path.relpath(path, REPO_ROOT))
        if SHOW_INLINE:
            display(Image(filename=path, width=420))
```

## 実験1: 潜在次元ごとの統計

```python
if "ex1" in EXPERIMENTS:
    show(plots.plot_ex1())
```

## 実験2: 潜在空間での移動量と出力の変化

```python
if "ex2" in EXPERIMENTS:
    for subset in EX2_SUBSETS:
        show(plots.plot_ex2(SEEDS, hyperparameter=HYPERPARAMETER, subset=subset))
```

## 実験3: 半径を固定したサンプリングとの比較

```python
if "ex3" in EXPERIMENTS:
    show(plots.plot_ex3(SEEDS, hyperparameter=HYPERPARAMETER))
```
